[1번] 70. Wilson’s Theorem  A number is prime if its only factors are 1 and itself. Write a program that determines whether a number is prime by using the theorem "The number n is a prime number if and only if n divides 

```python
def factorial(n: int) -> int:
    result = 1
    for i in range(1, n+1):
        result *= i
    return result

def isPrime(n: int) -> bool:
    if n < 2:
        return False
    return (factorial(n-1) + 1)%n == 0

number = int(input("Enter an integer greater than 1: "))

if (number > 1):
    if isPrime(number):
        print(f"{number} is a prime number.")
    else:
        print(f"{number} isn't prime number.")
```

[2번] 5. Alphabetical Order  The following words have three consecutive letters that are also consecutive letters in the alphabet: THIRSTY, NOPE, AFGHANISTAN, STUDENT. Write a program that accepts a word as input and determines whether or not it has three consecutive letters that are consecutive letters in the alphabet. The program should use a Boolean-valued function named isTripleConsecutive that accepts an entire word as input. Hint: Use the ord function.

```python
def isTripleConsecutive(target: str) -> bool:
    if len(target) < 3:
        return False
    
    for i in range(len(target) - 2):
        sliced = target[i:i+3]
        if (ord(sliced[1]) - ord(sliced[0])) == 1 and (ord(sliced[2]) - ord(sliced[0])) == 2:
            return True
    return False

word = input()
if isTripleConsecutive(word):
    print(f"{word} contains three successive letters in consecutive alphabetical order.")
else:
    print(f"{word} doesn't contain three successive letters in consecutive alphabetical order.")
```

[3번] 40. Months  The file SomeMonths.txt initially contains the names of the 12 months, one per line. Write a program that deletes all months from the file that do not contain the letter r. Your program should 1) print the original contents in the file, 2) delete all months from the file that do not contain the letter r, 3) print the updated contents in the file. In the sample output below, “...” is just for convenience, and your program should print all entries/lines in the file.

```python
file_name = input("1. Enter the file name: ")

print("2. Origianl contents in the file: ")
try:
    with open(file_name, 'r') as file:
        print(file.read())
except:
    print("Error while opening file")

# 근데 이거 대문자도 검사해야하는지 모르겟다 하지만!! r 이랬으니 소문자만 렛츠고
print("3. Deleting all months from the file that do not contain 'r'")
with_r = []
with open(file_name, 'r') as file:
    for line in file:
        if 'r' in line:
            with_r.append(line)

print("4. Updated contents in the file: ")
with open(file_name, 'w') as file:
    file.writelines(with_r)
with open(file_name, 'r') as file:
    print(file.read())
```

[4번] 46. File of Names  The file Names.txt contains a list of first names in alphabetical order. Write a program that requests a name from the user and inserts the name into the file in its proper location. If the name is already in the file, it should not be inserted. You should use set operations in your program.

```python
file_name = input("Enter the file name: ")
name_list = []
try:
    with open(file_name, 'r') as file:
        name_list = file.readlines()
except:
    print("Error while opening file")
name_list = set(name_list)

while True:
    first = input("Enter the first name to insert (or 'quit' to quit): ")
    if first == "quit":
        break
    first += "\n"
    if first in name_list:
        print(f"'{first.strip()}' cannot be inserted (already existing)")
    else:
        name_list.add(first)
        print(f"'{first.strip()}' is added to '{file_name}'")

name_list = sorted(list(name_list))
with open(file_name, 'w') as file:
    file.writelines(name_list)

print("Exiting the program.")
```
