### eval() 과 exec()
둘 다 매개변수로 문자열을 받음
 - eval은 "평가", 문자열(표현식)을 "실행"하고 결과를 "리턴"
 - exec은 "실행", 문자열(하나 이상의 문장)을 "실행", 리턴 없음

자료에 정리된 차이점
 - eval은 계산 가능한 식(또는 단순 숫자)을 입력 받고, 그 결과를 리턴
 - exec는 실행 가능한 명령문을 입력 받고, 결과를 리턴 하지 않음

– The eval function evaluates the expression to an integer or
floating-point number as appropriate
( eval 이 문자열 표현식을 실행 후 리턴하므로 형변환에 이용되기도 함)



```python
print( int("23") )
print( int(23.3) )
print( int("23.3") )

## int(float)는 float를 int로 바꾸는게 맞음
## 근데 int(str) 은 str이 int 처럼 생겼음을 가정함
## str을 float로 변환 후 int 로 바꾸는게 아님
```

### Format Method

```python
print('{0:10.2f}'.format(1234.5678))
print('{0:10,.2f}'.format(1234.5678))
print('{0:10,.3f}'.format(1234.5678))
print('{0:10,.2%}'.format(1234.5678))
print('{0:10,.3%}'.format(1234.5678))
print('{0:10,}'.format(12345678))
```

```python
# decimal (.) precision of 3 for a float
print('{0:.3f}'.format(1.0/3))
# fill with underscores (_) with the text centered (^)# to the width of 11
print('{0:_^11}'.format('hello'))
# keyword-based specifications
print('{name} wrote {book}'.format(name = 'Swaroop', book = 'A Byte of Python'))
```

```python
# • To specify some strings where no special processing 
#   such as escape sequences are handled
# – Specify a raw string by prefixing r or R to the string

print("New lines are indicated by \n.")
print(r"New lines are indicated by \n.")
```
