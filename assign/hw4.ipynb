[1번] 29. The following program crash if the user responds with “eight”. Rewrite the program using a try/except statement so that it will handle both types for responses as shown below.

```python
while True:
    n = input("Enter a nonzero integer: ")
    try:
        n = int(n)

    except:
    # except ValueError:
    # ValueError 를 명시해주는게 더 좋을지 아닐지 잘 모르겠다.
        print("You did not enter a nonzero integer. Try again.")
        continue

    if n != 0:
        reciprocal = 1 / n
        print("The reciprocal of {0} is {1:,.3f}".format(n, reciprocal))
        break
    else:
        print("You entered zero. Try again.")
```

[2번] 17. Matching Cards  52장으로 구성된 카드 덱이 총 2개 있다. 두 개의 덱을 각각 shuffle 한 뒤, 카드 덱의 위에서부터 각각 한 장씩 뽑아서 서로 같은지를 비교한다. 이 과정을 52장 모두에 대해서 시행하면 한번의 사이클이 끝난다. 한번의 사이클 동안 두 카드가 일치한 경우가 총 몇 번 발생했는지를 카운트한다. 총 10,000번의 사이클을 반복하고, 그 동안 두 카드가 일치한 경우가 몇 번 발생했는지를 계산한 뒤 이를 10,000으로 나눈 결과를 아래와 같이 출력하시오. 참고로, 랜덤의 영향으로 최종 결과는 아래와 다를 수 있음.

```python
from random import shuffle

def cycle() -> int:
    cnt = 0
    deck1 = [card for card in range(52)]
    deck2 = [card for card in range(52)]

    shuffle(deck1)
    shuffle(deck2)

    for i in range(52):
        if deck1[i] == deck2[i]:
            cnt += 1

    return cnt

total_cnt = 0
for _ in range(10000):
    total_cnt += cycle()

print(f"The average number of cards that matched was {total_cnt/10000:.3f}")
```

[3번] 7. Alphabetical Order  The following iterative function determines whether a list of lowercase words is in alphabetical order. Write the equivalent recursive function. Note that you can re-write only the lines in red.

```python
def isAlphaOrdered(L: str) -> bool:
    ## Determine whether list of lowercase words is in
    ## alphabetical order.
    if len(L) <= 1:
        return True
    if L[0] > L[1]:
        return False
    return isAlphaOrdered(L[1:])

test1 = "abcdef"
test2 = "ghijklmnzop"

print(f"For the string L= '{test1}' : {isAlphaOrdered(test1)}")
print(f"For the string L= '{test2}' : {isAlphaOrdered(test2)}")
```
